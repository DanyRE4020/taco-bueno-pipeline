```python
import pandas as pd

tablas = [
    "bronze_pos_raw",
    "silver_pos_lines"
]

resultados = []

for tabla in tablas:
    detalle = spark.sql(
        f"DESCRIBE DETAIL workspace.default.{tabla}"
    ).first().asDict()

    resultados.append({
        "tabla": tabla,
        "archivos_delta": detalle["numFiles"],
        "tamano_mib": round(
            detalle["sizeInBytes"] / (1024 ** 2),
            2
        ),
        "columnas_clustering": str(
            detalle.get("clusteringColumns", "No reportado")
        )
    })

display(pd.DataFrame(resultados))
```
