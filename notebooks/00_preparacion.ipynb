```python
CATALOGO = "workspace"
ESQUEMA = "taco_bueno"
VOLUMEN = "archivos_fuente"

# Crear el espacio donde guardaremos las tablas.
spark.sql(
    f"CREATE SCHEMA IF NOT EXISTS {CATALOGO}.{ESQUEMA}"
)

# Crear el espacio donde colocaremos los archivos de entrada.
spark.sql(
    f"CREATE VOLUME IF NOT EXISTS "
    f"{CATALOGO}.{ESQUEMA}.{VOLUMEN}"
)

RUTA_FUENTE = f"/Volumes/{CATALOGO}/{ESQUEMA}/{VOLUMEN}/"

print("Preparación completada.")
print("Espacio de tablas:", f"{CATALOGO}.{ESQUEMA}")
print("Ruta de archivos:", RUTA_FUENTE)
```
